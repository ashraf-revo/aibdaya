```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error
```

```python
plt.style.use('dark_background')
```

```python
n=20
x=np.arange(1,n)
y=np.arange(1,n)
```

```python
np.random.seed(42)
w=np.random.rand(1)[0]
b=np.random.rand(1)[0]

print(f"Initial weight: {w}, Initial bias: {b} with error {mean_squared_error(y, w*x+b)}")
```

```python
epochs = 1000
eta=.01
costvals=np.zeros(epochs)
```

```python
for i in range(epochs):
    yhat=w*x+b
    w=w-eta*(yhat-y)@x/len(x)
    b=b-eta*(yhat-y).mean()
    costvals[i]=mean_squared_error(y, yhat)
    
    if costvals[i] < .25:
        break

print(F"Best weight: {w}, Best bias: {b} at last cost {costvals[i]} with it={i}")
```

```python
costvals[:10]
```

```python
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.plot(x,y,label='Data')
plt.subplot(1,2,2)
plt.plot(range(i+1),costvals[:i+1], label='Cost')
plt.xlabel('Iterations')
plt.ylabel('Cost')
plt.title('Cost Function')
plt.legend()
plt.show()
```
