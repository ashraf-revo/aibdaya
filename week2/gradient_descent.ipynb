### Imports

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error
```

```python
np.random.seed(42)  # For reproducibility
```

```python
plt.style.use('dark_background')
```

### Batch Gradient Descent

```python
n=100
mean = [5, 15]
cov = [[19, 10], [10, 6]]
data = np.random.multivariate_normal(mean, cov, n)

x = data[:, 0]
y = data[:, 1]
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(x,y, color='blue', label='y=x',marker='o')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Linear Relationship')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

### Learning

```python
eta=.01
w=np.random.randn(1)[0]
b=np.random.randn(1)[0]

print(f'Initial weight: {w}, Initial bias: {b}')
```

```python
epochs=1000
costvals=[0] * epochs
history = []
history_rate=20
```

```python
history.append((w, b))
```

```python
for i in range(epochs):
    yhat = w * x + b
        
    w=w-eta*(yhat-y)@x/len(x)
    b=b-eta*(yhat-y).mean()
    
    if i % history_rate == 0:
        history.append((w, b))
    
    error=mean_squared_error(y, yhat)
    costvals[i]=error
    
    
```

```python
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)


wi,bi=history[0]
yhat=wi*x+bi
plt.plot(x,yhat, color='red', label='Initial Regression Line')

plt.scatter(x,y, color='blue', label='Data Points',marker='o')

for i in range(1,len(history)):
    wi,bi=history[i]
    yhat=wi*x+bi
    plt.plot(x,yhat, color='gray', alpha=0.5, label=None)

yhat=w*x+b
plt.plot(x,yhat, color='green', label='Final Regression Line')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Linear Regression Fit')
plt.grid( alpha=0.3)

plt.subplot(1, 2, 2)
plt.plot(costvals, color='orange', label='Cost over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Cost')
plt.title('Training Cost')
plt.legend()
plt.grid( alpha=0.3)

plt.show()
```

```python
print(f'Final weight: {w}, Final bias: {b} with cost ={costvals[-1]}')
```
