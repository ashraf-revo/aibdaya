## Visualization

```python
x=np.linspace(0,360,100)
y=np.sin(np.deg2rad(x))
z=np.cos(np.deg2rad(x))
```

```python
plt.style.use("dark_background")
plt.plot(x,y, label='sin(x)',marker='o', markersize=3, linestyle='--', color='cyan')
plt.plot(x,z, label='cos(x)', marker='o', markersize=3, linestyle='--', color='orange')
plt.title("sin vs cos wave")
plt.xlabel("x values")
plt.ylabel("sin(x) and cos(x)")
plt.legend()
```

```python
x=np.arange(1,50)
y=2*np.exp(-1*x)-1
plt.figure(figsize=(10, 6))
plt.plot(x, y,label="data=exp(-x)")
plt.xlabel("x")
plt.ylabel("data")
plt.legend()
plt.show()
```

```python
np.random.seed(42)  # For reproducibility
x=np.random.randn(50)
y=np.random.randn(50)

plt.figure(figsize=(10, 6))
plt.scatter(x, y,label="x")
plt.title("Scatter Plot of Random Data")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
class_x_1=np.random.randn(50)+2
class_y_1=np.random.randn(50)+2

class_x_2=np.random.randn(50)-2
class_y_2=np.random.randn(50)-2


plt.scatter(class_x_1, class_y_1, label="Class 1", color='blue', alpha=0.5)
plt.scatter(class_x_2, class_y_2, label="Class 2", color='red', alpha=0.5)
plt.axhline(0, color='blue')
plt.axvline(0, color='blue')
plt.title("Scatter Plot of Two Classes")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```
